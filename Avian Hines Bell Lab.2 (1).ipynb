*Exercise 1:* Write a function that creates an **n×n matrix** (a list of lists) representing the state of a tic‑tac‑toe game.

Use the integers:

- `0` = empty
- `1` = `"X"`
- `2` = `"O"`


````python
def make_board(n):
    board = []
    for _ in range(n):
        row = []
        for _ in range(n):
            row.append(0)
        board.append(row)
    return board
````

````python
def cell_to_char(value):
    if value == 0:
        return " "
    elif value == 1:
        return "X"
    else:
        return "0"
def draw_board(board):
    n = len(board)
    for row in range(n):
        print(' __' * n)
        line = "|"
        for col in range(n):
            line += " " + cell_to_char(board[row][col]) + " |"
        print(line)
    print(" ---" * n)
````

*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
def game_status(board):
    n = len(board)
    for r in range(n):
        first = board[r][0]
        if first !=0 and all(board[r][c] == first for c in range(n)):
            return first 
    for c in range (n):
        first = board[0][c]
        if first != 0 and all(board[r][c] == first for r in range(n)):
            return first
    first = board[0][0]
    if first != 0 and all(board[i][i] == first for i in range(n)):
        return first 
    first = board[0][n-1]
    if first != 0 and all(board[i][n-1-i] == first for i in range(n)):
        return first 
    for r in range(n):
        for c in range(n):
            if board[r][c] == 0:
                return -1 
    return 0
````

*Exercise 5:* Write a function that takes a game board, a player number, and `(row, col)` coordinates and places the correct mark (`"X"` or `"O"`) in that location.

Requirements:

- Only allow placing a mark in a previously empty location.
- Return `True` if the move was successful, and `False` otherwise.


````python
def place_move(board, player, row, col):
    if board[row][col] == 0:
        board[row][col] = player
        return True
    else:
        return False
````

````python
def draw_board_labeled(board):
    n = len(board)
    letters = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
    print("  " + "  ".join(letters[:n]))

    for r in range(n):
        print(" " + " ---" * n)
        row_str = f"{r+1} |"
        for c in range(n):
            row_str += f"{cell_to_char(board[r][c])} |"
        print(row_str)
    print(" " + " ---" * n)
````

*Exercise 7:* Write a function that takes a board, a player number, and a location string (as in Exercise 6), then uses your function from Exercise 5 to update the board.

````python
def place_move_from_string(board, player, location):
    col_letter = location[0].upper()
    row_number = location[1]
    col = ord(col_letter) - ord("A")
    row = int(row_number) - 1
    return place_move(board, player, row, col)
````

*Exercise 8:* Write a function that is called with a board and player number, takes input from the player using Python's `input()`, and modifies the board using your function from Exercise 7.

Keep asking for input until the player enters a valid location that results in a valid move.


````python
def player_turn(board, player):
    while True:
        move = input("Enter your move (e.g. A1, B2): ")
        success = place_move_from_string(board, player, move)
        
        if success:
            break
        else:
            print("Invalid move, try again.")
````

*Exercise 9:* Use all of the previous exercises to implement a full tic‑tac‑toe game:

- draw the board,
- repeatedly ask two players for a location,
- apply valid moves,
- check the game status until a player wins or the game is a draw.


````python
def play_game():
    board = make_board(3)
    current_player = 1

    while True:
        draw_board_labeled(board)

        player_turn(board, current_player)

        status = game_status(board)

        if status == 1:
            draw_board_labeled(board)
            print("Player 1 (X) wins!")
            break

        if status == 2:
            draw_board_labeled(board)
            print("Player 2 (O) wins!")
            break

        if status == 0:
            draw_board_labeled(board)
            print("It's a draw!")
            break

        current_player = 2 if current_player == 1 else 1
````

*Exercise 12:* Develop a version of the game where one player is the computer. This time, write a computer player using exhaustive search with a max depth parameter, similar to lecture.

````python
def computer_turn_depth(board, player, depth):
    if depth == 0 or game_status(board) != -1:
        return

    n = len(board)

    for r in range(n):
        for c in range(n):
            if board[r][c] == 0:
                board[r][c] = player
                if game_status(board) == player:
                    return
                board[r][c] = 0

    for r in range(n):
        for c in range(n):
            if board[r][c] == 0:
                board[r][c] = player
                computer_turn_depth(board, 1 if player == 2 else 2, depth - 1)
                return
````

*Exercise 13:* Make the 2 computer players play each-other for 10 games on a 3x3, then 4x4, then 5x5 grid. Set the max depth so that the games only take seconds. Measure the "smarter" player's win rate for each grid.

````python
import random

def available_moves(board):
    n = len(board)
    moves = []
    for r in range(n):
        for c in range(n):
            if board[r][c] == 0:
                moves.append((r, c))
    return moves

def minimax_limited(board, current_player, maximizing_player, depth):
    status = game_status(board)
    # game_status returns: -1 incomplete, 0 draw, 1 player1 wins, 2 player2 wins
    if status != -1 or depth == 0:
        if status == maximizing_player:
            return 10
        elif status == 0:
            return 0
        elif status == -1:
            return 0
        else:
            return -10

    moves = available_moves(board)
    if current_player == maximizing_player:
        best = -10**9
        for (r, c) in moves:
            board[r][c] = current_player
            score = minimax_limited(board, 1 if current_player == 2 else 2, maximizing_player, depth - 1)
            board[r][c] = 0
            if score > best:
                best = score
        return best
    else:
        best = 10**9
        for (r, c) in moves:
            board[r][c] = current_player
            score = minimax_limited(board, 1 if current_player == 2 else 2, maximizing_player, depth - 1)
            board[r][c] = 0
            if score < best:
                best = score
        return best

def computer_turn(board, player, depth):
    moves = available_moves(board)
    if not moves:
        return False

    # If depth is 0, just play random (fast “dumb”)
    if depth <= 0:
        r, c = random.choice(moves)
        board[r][c] = player
        return True

    best_score = -10**9
    best_moves = []

    for (r, c) in moves:
        board[r][c] = player
        score = minimax_limited(board, 1 if player == 2 else 2, player, depth - 1)
        board[r][c] = 0

        if score > best_score:
            best_score = score
            best_moves = [(r, c)]
        elif score == best_score:
            best_moves.append((r, c))

    r, c = random.choice(best_moves)
    board[r][c] = player
    return True

def play_computer_vs_computer(n, depth_p1, depth_p2):
    board = make_board(n)
    current = 1

    while True:
        status = game_status(board)
        if status != -1:
            return status  # 0 draw, 1 or 2 winner

        if current == 1:
            computer_turn(board, 1, depth_p1)
            current = 2
        else:
            computer_turn(board, 2, depth_p2)
            current = 1

def win_rate_smarter_player(n, smart_depth, dumb_depth, games=10):
    smart_wins = 0

    for _ in range(games):
        # Smarter player is Player 1
        result = play_computer_vs_computer(n, smart_depth, dumb_depth)
        if result == 1:
            smart_wins += 1

    return smart_wins / games

# ---------- RUN THE EXPERIMENT (10 games each) ----------
# Adjust these depths if it feels slow:
depths = {
    3: (4, 1),  # (smart_depth, dumb_depth)
    4: (2, 0),
    5: (1, 0),
}

for n in [3, 4, 5]:
    smart_depth, dumb_depth = depths[n]
    rate = win_rate_smarter_player(n, smart_depth, dumb_depth, games=10)
    print(f"{n}x{n} | smart depth={smart_depth}, dumb depth={dumb_depth} | smart win rate = {rate:.2f}")
````
