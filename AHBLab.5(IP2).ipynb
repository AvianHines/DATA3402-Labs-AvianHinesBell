```python
class matrix:

    def __init__(self, n=None, m=None, values=None):
        # init from list-of-lists
        if values is not None:
            self.data = values
            self.n = len(values)
            self.m = len(values[0])
        else:
            self.n = n
            self.m = m
            self.data = [[0.0 for j in range(m)] for i in range(n)]

    def __getitem__(self, index):
        i, j = index
        return self.data[i][j]

    def __setitem__(self, index, value):
        i, j = index
        self.data[i][j] = value

    def shape(self):
        return (self.n, self.m)

    def transpose(self):
        new_data = []
        for j in range(self.m):
            row = []
            for i in range(self.n):
                row.append(self.data[i][j])
            new_data.append(row)
        return matrix(values=new_data)

    def row(self, n):
        return matrix(values=[self.data[n]])

    def column(self, n):
        col = []
        for i in range(self.n):
            col.append([self.data[i][n]])
        return matrix(values=col)

    def to_list(self):
        return self.data

    def block(self, n0, n1, m0, m1):
        new_data = []
        for i in range(n0, n1):
            row = []
            for j in range(m0, m1):
                row.append(self.data[i][j])
            new_data.append(row)
        return matrix(values=new_data)

    def scalarmul(self, c):
        new_data = []
        for i in range(self.n):
            row = []
            for j in range(self.m):
                row.append(self.data[i][j] * c)
            new_data.append(row)
        return matrix(values=new_data)

    def add(self, N):
        if self.shape() != N.shape():
            raise ValueError("Matrix sizes do not match for add")
        new_data = []
        for i in range(self.n):
            row = []
            for j in range(self.m):
                row.append(self.data[i][j] + N.data[i][j])
            new_data.append(row)
        return matrix(values=new_data)

    def sub(self, N):
        if self.shape() != N.shape():
            raise ValueError("Matrix sizes do not match for sub")
        new_data = []
        for i in range(self.n):
            row = []
            for j in range(self.m):
                row.append(self.data[i][j] - N.data[i][j])
            new_data.append(row)
        return matrix(values=new_data)

    def element_mul(self, N):
        if self.shape() != N.shape():
            raise ValueError("Matrix sizes do not match for element-wise multiply")
        new_data = []
        for i in range(self.n):
            row = []
            for j in range(self.m):
                row.append(self.data[i][j] * N.data[i][j])
            new_data.append(row)
        return matrix(values=new_data)

    def matmul(self, N):
        if self.m != N.n:
            raise ValueError("Matrix sizes do not match for matmul")

        new_data = []
        for i in range(self.n):
            row = []
            for j in range(N.m):
                total = 0
                for k in range(self.m):
                    total += self.data[i][k] * N.data[k][j]
                row.append(total)
            new_data.append(row)
        return matrix(values=new_data)

    def equals(self, N):
        return self.data == N.data


    def __add__(self, N):
        return self.add(N)

    def __sub__(self, N):
        return self.sub(N)

    def __mul__(self, other):
        # scalar or matrix multiplication
        if type(other) == int or type(other) == float:
            return self.scalarmul(other)
        else:
            return self.matmul(other)

    def __rmul__(self, other):
        # scalar * M
        return self.__mul__(other)

    def __eq__(self, N):
        return self.equals(N)

def constant(n, m, c):
    return matrix(values=[[float(c) for j in range(m)] for i in range(n)])

def zeros(n, m):
    return constant(n, m, 0.0)

def ones(n, m):
    return constant(n, m, 1.0)

def eye(n):
    data = []
    for i in range(n):
        row = []
        for j in range(n):
            if i == j:
                row.append(1.0)
            else:
                row.append(0.0)
        data.append(row)
    return matrix(values=data)

def add(self, N):
    return self.add(N)

def sub(self, N):
    return self.sub(N)

def mul(self, other):
    if type(other) == int or type(other) == float:
            return self.scalarmul(other)
    return self.matmul(other)

def rmul(self, other):
    return self.__mul__(other)

def eq(self, N):
        return self.equals(N)

def ne(self, N):
        return not self.equals(N)
```
